```python
import nltk
from datasets import load_dataset
from transformers import AutoTokenizer

nltk.download("punkt", quiet=True)

MODEL_CHECKPOINT = "facebook/bart-large"
tokenizer = AutoTokenizer.from_pretrained(MODEL_CHECKPOINT)


dataset = load_dataset("knkarthick/samsum")

MAX_INPUT_LENGTH = 512
MAX_TARGET_LENGTH = 128

def preprocess_function(examples):

    model_inputs = tokenizer(
        examples["dialogue"],
        max_length=MAX_INPUT_LENGTH,
        padding="max_length",
        truncation=True
    )

    labels = tokenizer(
        text_target=examples["summary"],
        max_length=MAX_TARGET_LENGTH,
        padding="max_length",
        truncation=True
    )


    labels["input_ids"] = [
        [(label if label != tokenizer.pad_token_id else -100) for label in label_seq]
        for label_seq in labels["input_ids"]
    ]
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

# اعمال پیش‌پردازش به کل دیتاست
tokenized_datasets = dataset.map(
    preprocess_function,
    batched=True,
    remove_columns=dataset["train"].column_names
)
```

```python
from transformers import AutoModelForSeq2SeqLM
from peft import LoraConfig, get_peft_model, TaskType

# لود مدل اصلی
base_model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_CHECKPOINT)

# تعریف کانفیگ LoRA
lora_config = LoraConfig(
    task_type=TaskType.SEQ_2_SEQ_LM,
    r=8,
    lora_alpha=32,
    target_modules=["q_proj", "v_proj"],
    lora_dropout=0.05,
    bias="none"
)

# تبدیل مدل به ساختار PEFT
model = get_peft_model(base_model, lora_config)

# چاپ درصد پارامترهای قابل یادگیری برای گزارش فنی
model.print_trainable_parameters()
```

```python
import numpy as np
import evaluate
import nltk
from transformers import (
    DataCollatorForSeq2Seq,
    Seq2SeqTrainer,
    Seq2SeqTrainingArguments
)

# اطمینان از دانلود توکنایزر جملات
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

# 1. بارگذاری متریک ارزیابی ROUGE
rouge_metric = evaluate.load("rouge")

def compute_metrics(eval_pred):
    """
    محاسبه متریک‌های ROUGE با تبدیل ایمن شناسه‌ها و مدیریت مقادیر نامعتبر
    """
    predictions, labels = eval_pred

    if isinstance(predictions, tuple):
        predictions = predictions[0]

    # تبدیل مقادیر نامعتبر پیش‌بینی‌ها به pad_token
    predictions = np.where((predictions >= 0) & (predictions < tokenizer.vocab_size), predictions, tokenizer.pad_token_id)

    # تبدیل برچسب‌های -100 به pad_token
    labels = np.where((labels >= 0) & (labels < tokenizer.vocab_size), labels, tokenizer.pad_token_id)

    # تبدیل صریح به نوع داده صحیح پایتون برای جلوگیری از خطای تبدیل سرریز در Tokenizer
    predictions = predictions.astype(np.int64).tolist()
    labels = labels.astype(np.int64).tolist()

    # رمزگشایی ایمن
    decoded_preds = tokenizer.batch_decode(predictions, skip_special_tokens=True)
    decoded_labels = tokenizer.batch_decode(labels, skip_special_tokens=True)

    # تفکیک جملات برای محاسبه دقیق ROUGE-Lsum
    decoded_preds = ["\n".join(nltk.sent_tokenize(pred.strip())) for pred in decoded_preds]
    decoded_labels = ["\n".join(nltk.sent_tokenize(label.strip())) for label in decoded_labels]

    result = rouge_metric.compute(
        predictions=decoded_preds,
        references=decoded_labels,
        use_stemmer=True
    )

    return {k: round(v * 100, 2) for k, v in result.items()}

# 2. Data Collator پویا (Dynamic Padding) برای صرفه‌جویی شدید در مصرف VRAM
data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=model,
    pad_to_multiple_of=8  # بهینه‌سازی محاسبات ماتریسی روی معماری Tensor Cores
)

# 3. پیکربندی حرفه‌ای آموزش متناسب با کارت T4 Colab
training_args = Seq2SeqTrainingArguments(
    output_dir="./bart-samsum-lora-checkpoints",
    eval_strategy="epoch",            # ارزیابی دقیق در انتهای هر اپوک
    save_strategy="epoch",            # ذخیره چک‌پوینت همگام با ارزیابی
    learning_rate=2e-4,               # نرخ یادگیری استاندارد و تست‌شده برای آداپتورهای LoRA
    per_device_train_batch_size=4,    # بچ سایز ایمن برای جلوگیری از OOM
    per_device_eval_batch_size=4,
    gradient_accumulation_steps=4,    # بچ سایز مؤثر = 16 (پایداری کامل گرادیان)
    weight_decay=0.01,
    save_total_limit=1,               # نگه داشتن فقط بهترین مدل جهت مدیریت فضای دیسک
    num_train_epochs=3,               # ۳ اپوک برای همگرایی دیتای مکالمه کافیست
    predict_with_generate=True,       # فعال‌سازی فرآیند Autoregressive Generation در فاز ولیدیشن
    fp16=True,                        # کاهش ۵۰ درصدی مصرف حافظه با Mixed Precision
    logging_steps=50,                 # لاگ منظم برای رسم دقیق نمودار Loss
    generation_max_length=128,
    generation_num_beams=4,           # Beam Search بهینه برای کیفیت متن خروجی
    load_best_model_at_end=True,      # بارگذاری بهترین چک‌پوینت بر اساس eval_loss
    metric_for_best_model="eval_loss",
    greater_is_better=False,
    report_to="none"                  # غیرفعال‌سازی موقت سرویس‌های سنگین جانبی
)

# 4. آماده‌سازی موتور Trainer
trainer = Seq2SeqTrainer(
    model=model,
   args=training_args,
    train_dataset=tokenized_datasets["train"],
    eval_dataset=tokenized_datasets["validation"],
    processing_class=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)

# # 5. اجرای فرآیند آموزش
# print("=== Starting Training Pipeline ===")
# train_result = trainer.train()

# # چاپ خلاصه متریک‌های آموزش
# print("Training Metrics:", train_result.metrics)

# # 6. ذخیره‌سازی آداپتورهای سبک نهایی (کمتر از ۲۰ مگابایت)
# final_save_path = "./final_bart_lora_model"
# model.save_pretrained(final_save_path)
# tokenizer.save_pretrained(final_save_path)
# print(f"Adapters and tokenizer saved successfully to: {final_save_path}")
```

```python
import time
import torch
import gradio as gr
from peft import PeftModel, PeftConfig
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

device = "cuda" if torch.cuda.is_available() else "cpu"
adapter_path = '/content/drive/MyDrive/final_bart_lora_model'

# 1. بارگذاری بهینه مدل و توکنایزر
print("Loading model for Gradio deployment...")
peft_config = PeftConfig.from_pretrained(adapter_path)
base_model = AutoModelForSeq2SeqLM.from_pretrained(peft_config.base_model_name_or_path)
model = PeftModel.from_pretrained(base_model, adapter_path).to(device)
model.eval()
tokenizer = AutoTokenizer.from_pretrained(adapter_path)

# 2. تابع خلاصه‌سازی متصل به UI
def summarize_dialogue(text, num_beams=4, max_len=128):
    if not text.strip():
        return "Please provide a valid conversation.", "0.00s"

    inputs = tokenizer(text, return_tensors="pt", max_length=512, truncation=True)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    start_time = time.time()
    with torch.no_grad():
        summary_ids = model.generate(
            **inputs,
            num_beams=int(num_beams),
            max_length=int(max_len),
            min_length=15,
            length_penalty=2.0,
            no_repeat_ngram_size=3,
            early_stopping=True
        )
    latency = time.time() - start_time
    summary = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
    return summary, f"{latency:.2f}s"

# 3. تعریف اینترفیس گرافیکی
demo = gr.Interface(
    fn=summarize_dialogue,
    inputs=[
        gr.Textbox(
            lines=8,
            placeholder="Paste multi-turn chat/dialogue here...",
            label="Input Dialogue"
        ),
        gr.Slider(minimum=1, maximum=6, value=4, step=1, label="Beam Search Width"),
        gr.Slider(minimum=32, maximum=256, value=128, step=16, label="Max Summary Length")
    ],
    outputs=[
        gr.Textbox(lines=3, label="Generated Abstractive Summary"),
        gr.Label(label="Inference Latency")
    ],
    title="Dialogue Summarization Engine (BART-Large + LoRA)",
    description="Fine-tuned parameter-efficient summarizer trained on the SAMSum dialogue corpus. Optimized for low VRAM and zero-shot multi-speaker inference.",
    examples=[
        ["Hannah: Hey, do you have Betty's number?\nAmanda: Lemme check\nAmanda: Sorry, can't find it.\nAmanda: Ask Larry\nHannah: I'd rather you texted him\nAmanda: Just text him\nHannah: Alright, bye", 4, 128]
    ]
)

if __name__ == "__main__":
    demo.launch(share=True)
```

```python
import time
import torch
import gradio as gr
from peft import PeftModel, PeftConfig
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer

# 1. تنظیم دستگاه پردازشی (GPU در صورت وجود، در غیر این صورت CPU)
device = "cuda" if torch.cuda.is_available() else "cpu"

# اگر آداپتور را روی هاب آپلود کرده‌اید، نام مخزن را بگذارید؛ در غیر این صورت مسیر لوکال:
# مثال هاب: "YOUR_USERNAME/bart-large-samsum-lora"
MODEL_PATH = '/content/drive/MyDrive/final_bart_lora_model'

print(f"Loading model on {device}...")
peft_config = PeftConfig.from_pretrained(MODEL_PATH)
base_model = AutoModelForSeq2SeqLM.from_pretrained(peft_config.base_model_name_or_path)
model = PeftModel.from_pretrained(base_model, MODEL_PATH).to(device)
model.eval()

tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
print("Model and Tokenizer loaded successfully.")

# 2. پایپ‌لاین استنتاج با خطایابی
def summarize(dialogue_text, num_beams, max_length, min_length, length_penalty):
    if not dialogue_text or not dialogue_text.strip():
        return "⚠️ لطفاً یک گفتگوی معتبر وارد کنید.", "0.00s"

    try:
        inputs = tokenizer(
            dialogue_text,
            return_tensors="pt",
            max_length=512,
            truncation=True
        )
        inputs = {k: v.to(device) for k, v in inputs.items()}

        start_time = time.time()
        with torch.no_grad():
            summary_ids = model.generate(
                **inputs,
                num_beams=int(num_beams),
                max_length=int(max_length),
                min_length=int(min_length),
                length_penalty=float(length_penalty),
                no_repeat_ngram_size=3,
                early_stopping=True
            )
        latency = time.time() - start_time
        summary_text = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
        return summary_text, f"{latency:.2f}s"
    except Exception as e:
        return f"Error during generation: {str(e)}", "0.00s"

# نمونه‌های آماده برای تست سریع کارفرما یا کاربر
EXAMPLES = [
    [
        "Hannah: Hey, do you have Betty's number?\nAmanda: Lemme check\nAmanda: Sorry, can't find it.\nAmanda: Ask Larry\nAmanda: He called her last time we were at the park together\nHannah: I don't know him well\nAmanda: Don't be shy, he's very nice\nHannah: If you say so..\nHannah: I'd rather you texted him\nAmanda: Just text him 🙂\nHannah: Urgh.. Alright\nHannah: Bye\nAmanda: Bye bye",
        4, 128, 15, 2.0
    ],
    [
        "Benjamin: Hey guys, what are we doing with the keys today?\nHilary: I've got them. Whoever wants them can meet me at lunchtime or after\nElliot: I'm ok. We're meeting for drinks in the evening anyway\nBenjamin: I might pass by at lunchtime, take the keys and go take a nap.\nHilary: Sounds good. We'll be having lunch at La Cantina at 2 pm with some colleagues.\nBenjamin: See you there!",
        4, 128, 15, 2.0
    ]
]

# 3. طراحی اینترفیس واکنش‌گرا و مدرن
custom_css = """
#summary-box textarea { font-weight: 500; }
"""

with gr.Blocks(title="Dialogue Summarizer | BART-Large + LoRA", css=custom_css) as demo:
    gr.Markdown(
        """
        # 🎙️ سامانه خلاصه‌سازی هوشمند گفتگوها (BART-Large + LoRA)
        این مدل با استفاده از روش **PEFT (LoRA)** بر روی مجموعه داده **SAMSum** تنظیم دقیق شده و بدون از دست دادن اطلاعات گویندگان، خلاصه‌ای چکیده تولید می‌کند.
        """
    )

    with gr.Row():
        with gr.Column(scale=3):
            dialogue_input = gr.Textbox(
                label="متن گفتگو (Dialogue)",
                placeholder="متن مکالمه چندنفره را اینجا قرار دهید...",
                lines=10
            )
            with gr.Accordion("⚙️ تنظیمات پیشرفته تولید متن (Generation Parameters)", open=False):
                num_beams = gr.Slider(minimum=1, maximum=8, value=4, step=1, label="Beam Search Width")
                max_length = gr.Slider(minimum=32, maximum=256, value=128, step=8, label="حداکثر طول خلاصه (Max Length)")
                min_length = gr.Slider(minimum=10, maximum=64, value=15, step=5, label="حداقل طول خلاصه (Min Length)")
                length_penalty = gr.Slider(minimum=0.5, maximum=3.0, value=2.0, step=0.1, label="جریمه طول متن (Length Penalty)")

            btn_submit = gr.Button("تولید خلاصه ✨", variant="primary")

        with gr.Column(scale=2):
            summary_output = gr.Textbox(
                label="خلاصه استخراج‌شده (Summary)",
                lines=6,
                elem_id="summary-box"
            )
            latency_output = gr.Label(label="زمان پاسخ‌دهی (Latency)")

    gr.Examples(
        examples=EXAMPLES,
        inputs=[dialogue_input, num_beams, max_length, min_length, length_penalty],
        outputs=[summary_output, latency_output],
        fn=summarize,
        cache_examples=False
    )

    btn_submit.click(
        fn=summarize,
        inputs=[dialogue_input, num_beams, max_length, min_length, length_penalty],
        outputs=[summary_output, latency_output]
    )

if __name__ == "__main__":
    demo.launch(share=True)
```
